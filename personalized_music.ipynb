```python
# Personalized Music Recommendation

songs = {
    "ar rahman": ["Vinnaithaandi Varuvaayaa", "Hosanna", "Jai Ho"],
    "anirudh": ["Why This Kolaveri", "Arabic Kuthu", "Vaathi Coming"],
    "yuvan": ["Mun Paniya", "Oru Naalil", "Loosu Penne"]
}

favorite = input("Enter your favorite artist: ").strip().lower()

if favorite in songs:
    print("\nRecommended songs for you:")
    for song in songs[favorite]:
        print("🎵", song)
else:
    print("Sorry, no recommendations available.")
```
